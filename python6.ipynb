```python
#  Subset and Superset
u = {1, 2}
v = {1, 2, 3, 4}
print("Subset:", u.issubset(v))
print("Superset:", v.issuperset(u))

a={1,2,3,4}
b={1,2,3,4,5,6,7,8,9}
print("symmetric difference:",a^b)
print("is subset:",a.issubset(b))
print("is superset:",b.issuperset(a))
print("union:",a | b)
print("intersection:",a & b)
print("difference",b-a)
```

```python
# Remove Duplicates from List
list1 = [1, 2, 2, 3, 4, 4, 5]
unique_list = list(set(list1))
print("Unique List:", unique_list)

print(set(list1))

names=["shubham","nagesh","mayur","anurag","sanskar","shiva","yash","rutvik"]

shubh=set(names)

print(shubh)

shubh2=list(set(names))

print(shubh2)
```

```python
# Common Elements Between Lists
list2 = [1, 2, 3, 4]
list3 = [3, 4, 5, 6]
common = set(list2).intersection(set(list3))
print("Common Elements:", common)

name1=["shubham","mayur","sachin","kiran"]
name2=["anurag","rohit","anil","satish","shubham","mayur"]
name_list=set(name1).intersection(set(name2))
print(name_list)
```
