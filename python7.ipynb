```python
def add():
    a = 40
    b = 30
    c = a+b
    return c

print("The sum is:", add())

def multiply():
    a = 40
    b = 30
    c = a*b
    return c

print("The product is:", multiply())
```

```python
def add(a,b):
    return a+b

a = int(input("Enter a: "))    # will give error if you pass string here
b = int(input("Enter b: "))

print("Sum = ",add(a,b))

def multiply(a,b):
    return a*b

a = int(input("Enter a: "))
b = int(input("Enter b: "))

print("Product = ",multiply(a,b))
```

```python
# For readability, you can call function by writing the name of arguments

def find_distance(speed, time):
    print(speed, time)
    return speed * time
d = find_distance(speed=10, time=3)    # like this. this is readable. This is called KEYWORD ARGUMENTS.
print(d)

d = find_distance(time=3, speed=10)     # even if order is changed, no problem. This is called KEYWORD ARGUMENTS.
print(d)

d = find_distance(3,10)     # This is not the same as above (in this case, speed=3, time=10)
print(d)

def area(length, breadth):
    print(length, breadth)
    return length * breadth

a= area(length=10, breadth=5)
print(a)

a= area(breadth=5, length=10)
print(a)

a= area(10,5)     # This is not the same as above (in this case, length=10, breadth=5)
print(a)
```
