```python
# We can find the number is even or odd.

n = int(input(" Enter a number to check:"))

if n % 2 == 0:
  print("It's an even number.")
if n % 2 != 0:
  print("It's an odd number.")
if n == 0:
  print("The number is neither even nor odd. Rather you entered Zero.")

num = int(input(" Enter a number to check:"))
if num > 0:
  print("It's a positive number.")
if num < 0:
  print("It's a negative number.")
if num == 0:
  print("The number is neither positive nor negative. Rather you entered Zero.")
    
```

```python
# The else keyword catches anything which isn't caught by the preceding conditions.
# If the condition provided in the if statement is false, then the else statement will be executed.


a = 200
b = 33
if b > a:
  print("b is greater than a")
else:
  print("b is not greater than a")

a=45
b=11
if b > a:
  print("b is greater than a")
else:
  print("b is not greater than a")
```

```python
# Demonstration of elif
# elif is nothing but else if i.e. if you want to give condition to else block as well

x = 3
if x == 1:
  print('x is one')
elif x == 2:
  print('x is two')
elif x == 3:
  print('x is three')
else:
  print('x not found')

number = int(input("Enter a number: "))
if number <=10:
  print("The number is less than or equal to 10.")
elif number<=20:
  print("The number is less than or equal to 20.")
elif number<=30:
  print("The number is less than or equal to 30.")
else:
  print("The number is greater than 30.")
```
