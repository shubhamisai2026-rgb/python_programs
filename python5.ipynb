```python
# Simple example

company = {
  "name": "Tesla",              # this is called dictionary iteam. This dictionary has 3 items.
  "founder": 'Elon Musk',
  "established": 2003
}
print(company)

student={
    "name":"shubham",
    "roll_no":123,
     "class":"12th",
     "hobbies":"cricket"
}
print(student)
```

```python
# To access specific dictionary iteam, call it using its key

# Simple example

company = {
  "name": "Tesla",              # this is called dictionary iteam. This dictionary has 3 items.
  "founder": 'Elon Musk',
  "established": 2003
}
print(company)

student={
    "name":"shubham",
    "roll_no":123,
     "class":"12th",
     "hobbies":"cricket"
}
print(student)
print(company['name'])
print(student["name"])
print(student["roll_no"])
print(student["class"])
print(student["hobbies"])
```

```python
# To get length of dictionary

print(len(company))    # 3 items
print(len(student))
```

```python
# To change value of an item, just override the value as follows

company['name'] = 'TESLAA'
print(company)

student["hobbies"]="football"
print(student)

student["roll_no"]=254
print(student)

```

### Dictionary Functions

```python
# dict supports lot of functions

help(dict)
```

```python
# Another way to get item

print(company.get('name'))   # get() is predefined function for dict
print(company.get('namee'))  # namee is not defined, so we get None

print(student["name"])
print(student.get("name"))
print(student.get("roll_no"))
print(student.get("hobbies"))
print(student.get("class"))
```

```python
# keys() and values() functions gives us all the keys and  values

print(company.keys())
print(company.values())

print(student.keys())
print(student.values())
```

```python
# Let's add new key value pair to dictioary and see if we get the updated list with .key()

company['newkey'] = 'newval'
print(company.keys())
print(company.values())

student["subject"]="science"
print(student)
print(student.keys())
print(student.values())
```

```python
# TO delete an item, use pop

student = {
    'name' : 'John',
    'age' : 12,
    'class' : 7
}

print(student)

student.pop('class')    # use pop function and pass key
print(student)

college={
    "name":"NYNC",
    "principal":"shelke",
    "estamblish":2034,
    "chairman":"shubham_isai"

}
print(college)

college.pop("name")
print(college)

college.pop("chairman")
print(college)
```

```python
# del can also be used to delete an item

student = {
    'name' : 'John',
    'age' : 12,
    'class' : 7
}

print(student)

del student["name"]     # syntax -> del dict_name['key']
print(student)

state={
    "name":"maharashtra",
    "chief_minister":"shubham_isai",
     "districts":36,
     "capital":"mumbai"
}
print(state)

del state["name"]
print(state)
```

```python
# To delete complete dictionary, simply use del
# del student    # Once deleted you can't delete again.
```

```python
# clear() clears the dictionary without deleting the defined object. The dictionary just becomes empty (clear)

student = {
    'name' : 'John',
    'age' : 12,
    'class' : 7
}

print(student)
student.clear()
print(student)

state.clear()
print(state)
```

```python
# Let's first define dictionary in a different way

new = dict({'name': 'Apple', 'year': 1975, 'founder': 'Steve Jobs and Steve Wozniak'})     # dict is special keyword, wrap your dictionary within dict()

print(new)
print(type(new))

book=dict({"name":"mindset of indians","writter":"shubham isai","publish":2030})
print(book)
print(type(book))

book["state"]="maharashtra"
print(book)
```

```python
# simple for loop
for item in new:              # this loops over only keys
  print(item)


for item in book:
 print(item)
```

```python
# simple for loop
for item in new:              # this loops over only keys
  print(item, new[item])      # for each key, let's access its value with new[keyname]

for item in book:
  print(item,book[item])
```

```python
# Shorter way to do the same is to use .items() function

for item, val in new.items():
  print(item, val)

for item in book:
  print(item,book[item])
```

```python
# Loop through only keys

for key in new.keys():
  print(key)

for key in book:
  print(key)
```

```python
# Loop through only values

for xyz in new.values():
  print(xyz)

for xyz in book.keys():
  print(xyz)
for xyz in book.values():
  print(xyz)

print(book.keys())
print(book.values())
```

```python
# You can sort dictionary

for key in sorted(new):     # sorted() sorts all keys alphabetically
  print(key)

for key in sorted(book):
  print(key,book[key])

```
