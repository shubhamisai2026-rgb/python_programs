```python
# Define a list

courses = ['ba', 'bcom', 'bsc', 'ma', 'mcom', 'msc']
print(courses)                  # Note that the output is in square bracket. Remember, square brackets denote list

subjects=['math','cs','it','physics','chemistry','biology','marathi','geography']
print(subjects)
```

```python
# It's a good practice to know data type of a variable, so we know what we are dealing with

print(type(courses))
print(type(subjects))
```

```python
# Access first element

print(courses[0])
print(subjects[3])
print(subjects[0])
```

```python
# Access third element

print(courses[2])
print(subjects[2])
```

```python
# Access first and second element

print(courses[0:2])
print(subjects[0:7])
```

```python
# Access 2nd to 5th element (i.e bcom bsc ma mcom)
# 2nd element = 1st index, 5th element = 4th index

print(courses[1:5])   # this is nothing but courses[1], courses[2], courses[3], courses[3]
# when using ranges, the output EXCLUDES the second number (hence, if we write index 5, then we get index 4th as max)
print(subjects[2:7])
```

```python
# Access last element (sometimes you don't know how long a list is, so a quick shortcut is)

print(courses[-1])   # -1 represents last
print(subjects[-1])
print(subjects[-2])
print(subjects[-3])
```

```python
print(courses[-2])   # -2 represents second last
print(subjects[-4])
print(subjects[-5])
```

```python
print(courses[-3:-1]) # print from thirdlast  to last (excluding last)
print(subjects[-4:-1])
```

```python
print(courses[-3:]) # print from thirdlast  to last (including last)
print(subjects[-7:])
print(subjects)
```

### List functions

```python
# Let's first check what all functions list supports
# ignore the functions that start with __

help(list)       # list is a special and reserved keyword in python. It should NOT be used as variablename
```

```python
# Let's start with the first function from the list - append()
# Let's check what append does and its definition.
# In the output, ignore SELF. SELF is nothing but the variable that is calling the fuction.

help(list.append)
```

```python
# this is the same as above
help(courses.append)    # see it's not saying self here, because we have already passed the SELF (i.e. courses in this case) while asking for help
```

```python
# Let's use append to append an element to our list.

courses.append('BE')
print(courses)

subjects.append("history")
print(subjects)

subjects.append("psychology")
print(subjects)
```

```python
# To insert an element somewhere in the middle, say 3rd index

courses.insert(3, 'ME')           # i.e. insert requires position and value
print(courses)

subjects.insert(4,"jiology")
print(subjects)
```

```python
# You can append another variable as well

c = 'B.Pharm'
courses.append(c)
print(courses)

d="tech"
subjects.append(d)
print(subjects)
```

```python
# You can append a list within a list

new = ['11th', '12th']   # Defining a new list
courses.append(new)      # Appending new list to the old list
print(courses)           # Note that the list itself got inserted into the list. That list constitutes one element and not two.

new2=['english','sanskrut']
subjects.append(new2)
print(subjects)
```

```python
print(courses[-1])
print(subjects[-1])
```

```python
# Note, in the above example, the list got inserted as a list only and not as individual items.
# To add individual list items to the original list, use the function extend

help(courses.extend)
```

```python
# The last 4 words are important in above definition - ELEMENTS FROM THE ITERABLE. ITERABLE in this case is thew new list that we are passing (see the first argument being passed to the function definiion)
# In other words, extend actually unpacks the other list and adds items to the original list instead of creating a nested list by simply inserting one list into the other

# Let's check
courses.extend(new)
print(courses)
subjects.extend(new2)
print(subjects)

```

Now the new list got inserted NOT as a list, but as separate elements

```python
# To remove last entry of list, use pop()

courses.pop()
print(courses)
subjects.pop()
print(subjects)
```

```python
# To remove specific item, use remove()


subjects.remove("marathi")
print(subjects)
```

```python
# Let's define a simple for loop

for item in countries:            # item could be changed by any other random name.
  print(item)

for i in districts:
  print(i)

for j in fruits:
  print(j)

for k in numbers:
  print(k)
```

Basically, we are telling python that, for every element in **countries**, print that element

```python
for x in countries:             # x can be changed to any other word
  print(x)
for item in districts:
  print(item)
```

```python
# Note that : is required at the end of for line. Also, the next code block needs to have 4 indentation (i.e. blank space).
# Check the difference between these two examples


for i in districts:
  print(i)
print("outside of the for loop")

print("---------------------------------")
for i in districts:
  print(i)
  print("inside for loop")
```

```python
# Let's spice up the things little bit

for item in countries:
  print(item, len(item))          # printing each item and its length

for item in districts:
  print(item,len(item))
```

```python
for item in countries:
  print(item, item.upper(), len(item))          # Let's capitalize as well
for item in districts:
  print(item,item.upper(),len(item))
```

```python
for item in countries:
  print(f'{item}, {item.upper()}, {len(item)}')          # Let's use f string for printing. Remember inside {} you have to pass variable (or python function or python code)

for item in districts:
  print(f'{item},{item.upper()},{len(item)}')
```

```python
for item in countries:
  print(f'{item} --- {item.upper()} --- {len(item)}')  # f string allows to easily combine our print strings and python codes
for item in districts:
  print(f'{item}-----{item.upper()}-------{len(item)}')
```

```python
# Use enumerate() function to get index of element.

for id, item in enumerate(countries):   # enumerate returns index along with the actual item, hence we need to catch it explicitly.
  print(id, item)
for id,item in enumerate(districts):
  print(id,item)
for id,item in enumerate(fruits):
  print(id,item)
```

```python
for index, item in enumerate(countries):   # id can be changed to anyother thing
  print(index, item)
for index,item in enumerate(districts):
  print(index,item)
```

```python
# let's try to create a list using for loop and append()

cubes = []    # first define a blank list
for i in range(5):       # you can explore what range(5) gives. In short, it returns numbers upto 5.
  cubes.append(i ** 3)
print(cubes)

squares=[]
for i in range(10):
  squares.append(i**2)
print(squares)

quadratic=[]
for i in range(10):
  quadratic.append(i**4)
print(quadratic)
```

```python
# To deconstruct what is exactly happening in the above code, you can use print() like this

cubes = []
for i in range(5):
  print("i = ", i)
  cubes.append(i ** 3)
  print("Number appended (i ** 3) = ", i ** 3)
  print("Current status of cubes list = ", cubes)
print()
print('Outside for loop')
print(cubes)

quadratic=[]
for i in range(10):
  print("i=",i)
  quadratic.append(i**4)
  print("number appended (i**4)=",i**4)
  print('current ststus of quadratic list=',quadratic)
 
print()
print("outside the for loop")
print(quadratic)


```
